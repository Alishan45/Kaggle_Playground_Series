## Data Description
Synthetic dataset for the binary classification of poisonous mushrooms with a mixture of continuous and categorical features, including several features with significant missing data.

### Columns in the Dataset

- **class**: Indicator of whether the mushroom is poisonous (p) or edible (e)
- **cap-diameter**: Diameter of the mushroom cap
- **cap-shape**: Shape of the mushroom cap
- **cap-surface**: Surface texture of the mushroom cap
- **cap-color**: Color of the mushroom cap
- **does-bruise-or-bleed**: Indicator of whether the mushroom bruises or bleeds
- **gill-attachment**: Attachment type of the mushroom gills
- **gill-spacing**: Spacing of the mushroom gills
- **gill-color**: Color of the mushroom gills
- **stem-height**: Height of the mushroom stem
- **stem-width**: Width of the mushroom stem
- **stem-root**: Root type of the mushroom stem
- **stem-surface**: Surface texture of the mushroom stem
- **stem-color**: Color of the mushroom stem
- **veil-type**: Type of the mushroom veil
- **veil-color**: Color of the mushroom veil
- **has-ring**: Indicator of whether the mushroom has a ring
- **ring-type**: Type of the mushroom ring
- **spore-print-color**: Color of the mushroom spore print
- **habitat**: Habitat where the mushroom is found
- **season**: Season when the mushroom is found

# Data distribution w.r.t to class + Corr analysis

```python
import warnings

def plot_continuous_distribution(df, features, target, sample_size=1000000):
    num_features = len(features)

    df_sample = df.groupby(target, group_keys=False).apply(lambda x: x.sample(min(len(x), sample_size // df[target].nunique()), random_state=42))

    for i in range(0, num_features, 2):
        fig, axes = plt.subplots(2, 4, figsize=(20, 10))

        for j in range(2):
            if i + j < num_features:
                feature = features[i + j]

                with warnings.catch_warnings():
                    warnings.simplefilter(action='ignore', category=FutureWarning)

                    sns.histplot(df_sample, x=feature, hue=target, kde=True, ax=axes[0, j*2], palette=dict(zip(unique_labels, palette)))
                    axes[0, j*2].set_title(f'{feature} - Histogram & KDE')
                    axes[0, j*2].grid(True)

                    sns.violinplot(x=target, y=feature, data=df_sample, ax=axes[0, j*2+1], palette=dict(zip(unique_labels, palette)))
                    axes[0, j*2+1].set_title(f'{feature} - Violin Plot')
                    axes[0, j*2+1].grid(True)

                stats.probplot(df_sample[feature], dist="norm", plot=axes[1, j*2])
                axes[1, j*2].get_lines()[1].set_color('red')
                axes[1, j*2].set_title(f'{feature} - QQ Plot')
                axes[1, j*2].grid(True)

                desc_stats = df_sample[feature].describe()
                skewness = df_sample[feature].skew()
                kurtosis = df_sample[feature].kurt()
                stats_text = (f'{feature} - Descriptive Statistics:\n{desc_stats}\n'
                              f'Skewness: {skewness:.2f}, Kurtosis: {kurtosis:.2f}')
                axes[1, j*2+1].text(0.5, 0.5, stats_text, ha='center', va='center', fontsize=12,
                                    transform=axes[1, j*2+1].transAxes, bbox=dict(facecolor='white', alpha=0.5))
                axes[1, j*2+1].axis('off')
            else:
                for k in range(2):
                    axes[0, j*2+k].axis('off')
                    axes[1, j*2+k].axis('off')
        plt.tight_layout(pad=3.0)
        plt.show()

# Plot continuous distribution
plot_continuous_distribution(train, continuous_cols, 'class')

# Plot correlation heatmap
corr_matrix = train[continuous_cols].corr()
fig, ax = plt.subplots(figsize=(6, 6))
sns.heatmap(corr_matrix, annot=False, cmap='viridis', center=0,
            square=True, linewidths=.5, cbar_kws={"shrink": .5}, ax=ax)
ax.set_title('Correlation Heatmap of All Features', fontsize=18)
fig.tight_layout()
plt.show()

# Print correlation matrix
print(corr_matrix)
```

```python
def pairplot(df, features, target, sample_size=1000000):
    df_sample = df.groupby(target, group_keys=False).apply(lambda x: x.sample(min(len(x), sample_size // df[target].nunique()), random_state=42))

    sns.set(style="whitegrid", palette=palette)

    with warnings.catch_warnings():
        warnings.filterwarnings("ignore", category=FutureWarning)
        pairplot = sns.pairplot(df_sample, vars=features, hue=target,
                                diag_kind='kde',
                                plot_kws={'alpha': 0.7, 's': 50, 'edgecolor': 'k'},
                                diag_kws={'fill': True},
                                palette='Set1')

    pairplot.fig.suptitle('Pairplot of Continuous Features', fontsize=16, y=1.02)
    for ax in pairplot.axes.flatten():
        ax.set_xlabel(ax.get_xlabel(), fontsize=10)
        ax.set_ylabel(ax.get_ylabel(), fontsize=10)
        ax.tick_params(axis='both', which='major', labelsize=8)

    handles = pairplot._legend.legendHandles
    labels = list(pairplot._legend.texts)
    pairplot._legend.remove()
    pairplot.fig.legend(handles=handles, labels=[label.get_text() for label in labels], title=target, loc='upper right', bbox_to_anchor=(1.15, 1))

    pairplot.fig.set_size_inches(12, 12)
    pairplot.fig.subplots_adjust(top=0.9)
    pairplot.fig.tight_layout(pad=2.0)
    plt.show()

pairplot(train, continuous_cols, "class")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

def plot_categorical_distribution_with_pie(df, features, target):
    num_features = len(features)
    unique_labels = df[target].unique()
    palette_1 = sns.color_palette("Paired")

    def autopct_format(values):
        def inner_autopct(pct):
            return ('%1.1f%%' % pct) if pct > 0 else ''
        return inner_autopct

    for i in range(num_features):
        feature = features[i]
        fig, axes = plt.subplots(1, 2, figsize=(24, 10))

        # Count plot
        sns.countplot(x=df[feature], hue=df[target], palette=palette, ax=axes[0])
        axes[0].set_title(f'{feature} - Count Plot')
        axes[0].grid(True, axis='y')
        category_counts = df[feature].value_counts()
        xticks = axes[0].get_xticks()
        xticklabels = axes[0].get_xticklabels()

        new_labels = [label.get_text() if category_counts.get(label.get_text(), 0) > 0 else '' for label in xticklabels]
        axes[0].set_xticklabels(new_labels, rotation=90, ha='right')

        for tick in axes[0].get_xticklabels():
            tick.set_horizontalalignment('right')

        p_counts = df[df[target] == 'p'][feature].value_counts()
        e_counts = df[df[target] == 'e'][feature].value_counts()
        axes[0].plot(p_counts.index, p_counts.values, label='p', color='red', linestyle='--')
        axes[0].plot(e_counts.index, e_counts.values, label='e', color='green', linestyle='--')

        axes[0].legend(title='Class', labels=['Poisonous', 'Edible'])

        feature_counts = df[feature].value_counts()
        explode = [0.1] * len(feature_counts)  

        wedges, texts, autotexts = axes[1].pie(
            feature_counts, labels=None, autopct=autopct_format(feature_counts), startangle=140,
            colors=palette_1[:len(feature_counts)], explode=explode, wedgeprops={'edgecolor': 'white', 'linewidth': 1}
        )
        axes[1].set_title(f'{feature} - Pie Chart')
        bbox_props = dict(boxstyle="square,pad=0.3", edgecolor="black", linewidth=1, facecolor="white", alpha=0.8)
        kw = dict(arrowprops=dict(arrowstyle="-"), bbox=bbox_props, zorder=0, va="center")
        
        for i, (wedge, count, pct) in enumerate(zip(wedges, feature_counts, autotexts)):
            angle = (wedge.theta2 - wedge.theta1) / 2. + wedge.theta1
            x = np.cos(np.radians(angle))
            y = np.sin(np.radians(angle))
            horizontalalignment = {-1: "right", 1: "left"}[int(np.sign(x))]
            connectionstyle = "angle,angleA=0,angleB={}".format(angle)
            kw["arrowprops"].update({"connectionstyle": connectionstyle})
            axes[1].annotate(f"{feature_counts.index[i]}: {count} ({pct.get_text()})", 
                             xy=(x, y), xytext=(1.5*np.sign(x), 1.5*y),
                             horizontalalignment=horizontalalignment, **kw)

        plt.tight_layout(pad=3.0)
        plt.show()

plot_categorical_distribution_with_pie(train, categorical_cols, "class")
```

<a id="Encoding + Prep"></a>
# <div style="text-align:center; border-radius:15px 50px; padding:7px;font-weight:bold; color:black; margin:0; font-size:110%; background-color:white; border:2px solid black"><b>Encoding + Prep</b></div>



```python
encoder = OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1)
train[categorical_cols] = encoder.fit_transform(train[categorical_cols])
```

```python
y = train['class']
X = train.drop('class',axis=1)

class_mapping = {
    'e':0,
    'p':1
}
y = y.map(class_mapping)
# label_encoder = LabelEncoder()
# y = label_encoder.fit_transform(y)
```

```python
mcc_score = make_scorer(matthews_corrcoef)
models = {
    "XGB": XGBClassifier(random_state=42, tree_method = "hist", device = "cuda"),
    "LGBM": LGBMClassifier(random_state=42, verbose=0, device='gpu'),
    "HistGradientBoosting": HistGradientBoostingClassifier(random_state=42),
    "Cat": CatBoostClassifier(random_state=42, verbose=0)
}

kf = KFold(n_splits=3, shuffle=True, random_state=1)

results = {}

for name, model in models.items():
    cv_results = cross_val_score(model, X, y, cv=kf, scoring=mcc_score,n_jobs=-1)
    results[name] = cv_results
    print(f"{name} - Matthews Corr Coef: {cv_results.mean():.4f} (+/- {cv_results.std():.4f})")


results_df = pd.DataFrame(results).T
results_df.columns = ['Fold 1', 'Fold 2', 'Fold 3']
results_df['Mean Accuracy'] = results_df.mean(axis=1)
results_df['Std Dev'] = results_df.std(axis=1)

print("\nSummary of results:")
print(results_df)
```

<a id="Models validation with KFold 3"></a>
# <div style="text-align:center; border-radius:15px 50px; padding:7px;font-weight:bold; color:black; margin:0; font-size:110%; background-color:white; border:2px solid black"><b>Models validation with KFold 3</b></div>


```python
from sklearn.inspection import permutation_importance
for name, model in models.items():
    if name not in ["HistGradientBoosting", "Cat"]:
        model.fit(X, y)
fig, axes = plt.subplots(nrows=2, ncols=1, figsize=(15, 15))

for ax, (name, model) in zip(axes, models.items()):
    if name == "Cat":
        break;
    elif name == "HistGradientBoosting":
        break;
    else:
        importances = model.feature_importances_
        indices = np.argsort(importances)[::-1]
        feature_names = X.columns[indices]
        importances = importances[indices]
        sns.barplot(x=importances, y=feature_names, ax=ax, color='#0099ff', edgecolor='black')

    ax.set_xlabel('Importance', fontsize=14)
    ax.set_ylabel('Feature', fontsize=14)
    ax.set_title(name, fontsize=16, fontweight='bold')
    ax.spines['right'].set_visible(False)
    ax.spines['top'].set_visible(False)

plt.tight_layout()
plt.show()
```

# Training best model from validation list

```python
xgb = XGBClassifier(objective= 'binary:logistic',tree_method = "hist", device = "cuda", random_state=42)

xgb.fit(X, y)
```
